```python
import pandas as pd
url="https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(url)
```

```python
df.groupby("Pclass")["Survived"].mean()
```

```python
df.groupby(["Pclass","Sex"])["Survived"].mean()
```

```python
rates=df.groupby(["Pclass","Sex"])["Survived"].mean()
rates.idxmin()
rates.min()
```

```python
max_price=df.groupby("Embarked")["Fare"].agg(["count", "mean", "max"])
print("max price:",max_price)
```

```python
ports = pd.DataFrame({
    "Embarked": ["S", "C", "Q"],
    "Port": ["Southampton", "Cherbourg", "Queenstown"],
})
full = df.merge(ports, on="Embarked", how="left")
print(full[["Name", "Embarked", "Port"]].head())
```
