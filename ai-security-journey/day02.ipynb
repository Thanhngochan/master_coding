```python
import time
nums = list(range(1_000_000))
arr = np.arange(1_000_000)

t = time.time(); [x * 2 for x in nums]; t_list = time.time() - t
t = time.time(); arr * 2;               t_np = time.time() - t
print(f"list: {t_list:.4f}s | numpy: {t_np:.4f}s | {round(t_list / t_np)}x faster")
```
